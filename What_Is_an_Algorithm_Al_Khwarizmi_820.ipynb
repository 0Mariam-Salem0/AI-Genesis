## 📖 Part 1: The story

Around **820 CE**, the caliph al-Ma'mun ran the **House of Wisdom** in Baghdad, the greatest centre of science of its time. One of its scholars was **Muhammad ibn Musa al-Khwarizmi**, from the region of Khwarazm in Central Asia.

He wrote two books that changed the world:

1. **A book on Indian (Hindu-Arabic) numerals.** It explained how to calculate with the ten digits **0 to 9** using a *positional system*, where a digit's place gives its value, and **zero** holds an empty place. It also gave **step-by-step procedures** for adding, subtracting, multiplying and dividing.
2. ***Al-Kitab al-mukhtasar fi hisab al-jabr wal-muqabala*** (*The Compendious Book on Calculation by Completion and Balancing*). It showed how to solve equations **systematically**, with clear rules instead of tricks.

### 🔤 Two words from one man
- From **al-jabr** ("completion", one of his operations) comes the word **algebra**.
- In the **12th century**, his arithmetic book was translated into Latin, starting with *"Dixit Algorizmi"* ("Al-Khwarizmi said"). The Latin form of his name, **Algoritmi**, became **algorism** (decimal calculation), and later **algorithm**, meaning *any* step-by-step procedure.

So when you write code, you use a word that comes from a person's name. 🙂

### 🌍 What happened next
- **1202:** Fibonacci's *Liber Abaci* spread the Hindu-Arabic numerals across Europe.
- **~300 BCE** (much earlier): Euclid already had a famous algorithm (for the greatest common divisor). Al-Khwarizmi gave us the **idea of algorithms as a general method**.
- **1936:** Alan Turing and Alonzo Church made the idea of "algorithm" mathematically precise.

### 🗺️ Where this fits
```
Aristotle (350 BCE) → Al-Khwarizmi (820) → Leibniz: binary (1703) → Lovelace: first program (1843) → Turing (1936)
```

## 🧠 Part 2: The ideas in plain words

### What is an algorithm?
An **algorithm** is a **precise list of steps** that turns an **input** into an **output**. A cooking recipe is close, but a real algorithm must be *much* more exact. Computer scientists ask for five properties:

| Property | Meaning | Example (adding two numbers on paper) |
|---|---|---|
| **Input** | It starts from some given values | the two numbers |
| **Output** | It produces a result | their sum |
| **Definiteness** | Every step is exact, with no guessing | "add the two digits and the carry" |
| **Finiteness** | It stops after a finite number of steps | it ends when the digits run out |
| **Effectiveness** | Every step is simple enough to actually do | adding two single digits |

> 💡 *"Add some salt until it tastes right"* is **not** an algorithm. It is not definite.

### A program vs an algorithm
An **algorithm** is the idea (the recipe). A **program** is that idea written in a programming language so a computer can run it.

### Why efficiency matters
Two algorithms can both be correct, yet one takes **3 steps** and the other **3 million**. Counting the steps is how we compare them. Later notebooks on search and sorting rely on this idea all the time.

## ➕ Step 3: Al-Khwarizmi's decimal arithmetic
> 💡 Adding on paper is an algorithm. You work **digit by digit, right to left**, and carry when a column reaches 10.
> Our function follows exactly these steps and records every one, so you can see the algorithm run: `478 + 965`.

````python
# The school addition algorithm, step by step
def add_with_carries(a, b):
    x, y = str(a)[::-1], str(b)[::-1]                    # reverse so we start from the units digit
    n = max(len(x), len(y))
    carry, out, rows = 0, [], []
    for i in range(n):
        da = int(x[i]) if i < len(x) else 0
        db = int(y[i]) if i < len(y) else 0
        total = da + db + carry
        rows.append({"column": ["units", "tens", "hundreds", "thousands", "ten-thousands"][i] if i < 5 else f"10^{i}",
                     "digit of a": da, "digit of b": db, "carry in": carry,
                     "sum": total, "write down": total % 10, "carry out": total // 10})
        out.append(str(total % 10))
        carry = total // 10                              # carry 1 if the column reached 10 or more
    if carry:
        out.append(str(carry))
    return int("".join(out[::-1])), pd.DataFrame(rows)

result, steps = add_with_carries(478, 965)
assert result == 478 + 965
print(f"478 + 965 = {result}")
steps
````

## 🔢 Step 4: Why position and zero matter
> 💡 The big idea of the numerals in his book: a digit's **place** decides its value, and **0** says "this place is empty".
> `4305` means `4×1000 + 3×100 + 0×10 + 5×1`. The **same algorithm** that turns digits into a number works in **any base**. Let us convert a number into **binary** (base 2) with repeated division. This is the algorithm computers use, and it connects to Leibniz's binary system (1703), a later notebook.

````python
# Value of each digit in 4305, then decimal -> binary by repeated division by 2
digits = "4305"
rows = [{"digit": d, "place": 10 ** (len(digits) - 1 - i), "value": int(d) * 10 ** (len(digits) - 1 - i)}
        for i, d in enumerate(digits)]
place_table = pd.DataFrame(rows)
print(place_table.to_string(index=False), f"\nTotal = {place_table['value'].sum()}\n")

def to_binary(n):
    steps, bits = [], []
    while n > 0:                                         # keep dividing until nothing is left
        steps.append({"number": n, "÷ 2 = ": n // 2, "remainder": n % 2})
        bits.append(str(n % 2))
        n //= 2
    return "".join(bits[::-1]), pd.DataFrame(steps)      # read the remainders from the bottom up

binary, division_steps = to_binary(37)
assert binary == bin(37)[2:]
print(f"37 in binary = {binary}  (read the remainders from bottom to top)")
division_steps
````

## 📐 Step 5: His algebra, with a picture
> 💡 Al-Khwarizmi solved equations using **geometry**. His most famous example in the book:
>
> *"A square and ten roots equal thirty-nine dirhams"*, which we write as **x² + 10x = 39**.
>
> His method, **completing the square**:
> 1. Half of the roots: `10 / 2 = 5`
> 2. Square it: `5 × 5 = 25`
> 3. Add it to both sides: `39 + 25 = 64`
> 4. The square root: `√64 = 8`
> 5. Subtract the half: `8 − 5 = 3`. So **x = 3**.
>
> The picture shows *why*: a square with side `x`, plus two rectangles `5 × x` (that is the `10x`), plus a corner square `5 × 5`, forms one big square of area `39 + 25 = 64`, so its side is `8`.

````python
# Draw the completing-the-square picture for x^2 + 10x = 39  (x = 3)
x_val, half = 3, 5
fig, ax = plt.subplots(figsize=(6.4, 6.4))
ax.add_patch(Rectangle((0, 0), x_val, x_val, fc=GREEN, ec="white", lw=3))                  # the square x*x
ax.add_patch(Rectangle((x_val, 0), half, x_val, fc=GOLD, ec="white", lw=3))                # first 5 x x rectangle
ax.add_patch(Rectangle((0, x_val), x_val, half, fc=GOLD, ec="white", lw=3))                # second 5 x x rectangle
ax.add_patch(Rectangle((x_val, x_val), half, half, fc=ORANGE, ec="white", lw=3, alpha=0.8))  # corner square 5*5

ax.text(x_val / 2, x_val / 2, "x²\n(a square)", ha="center", va="center", color="white", fontweight="bold", fontsize=12)
ax.text(x_val + half / 2, x_val / 2, "5x", ha="center", va="center", color="white", fontweight="bold", fontsize=13)
ax.text(x_val / 2, x_val + half / 2, "5x", ha="center", va="center", color="white", fontweight="bold", fontsize=13)
ax.text(x_val + half / 2, x_val + half / 2, "25\n(the part we add)", ha="center", va="center", color="white", fontweight="bold", fontsize=12)
ax.text(x_val / 2, -0.45, "x = 3", ha="center", color=DARK); ax.text(x_val + half / 2, -0.45, "5", ha="center", color=DARK)
ax.set_xlim(-0.8, 8.8); ax.set_ylim(-1, 9.4); ax.set_aspect("equal"); ax.axis("off")
ax.set_title("x² + 10x = 39  →  add 25  →  a big square of area 64, side 8", color=DARK, fontweight="bold", fontsize=11)
plt.show()
````

## 🤖 Step 6: His procedure as code
> 💡 He gave rules for **six kinds of equations** (he did not use zero or negative numbers, so each kind is written with only positive amounts):
>
> | Type | In his words | In symbols |
> |---|---|---|
> | 1 | squares equal roots | x² = bx |
> | 2 | squares equal numbers | x² = c |
> | 3 | roots equal numbers | bx = c |
> | 4 | squares and roots equal numbers | x² + bx = c |
> | 5 | squares and numbers equal roots | x² + c = bx |
> | 6 | roots and numbers equal squares | bx + c = x² |
>
> We code the rules for **types 4 and 5** with the narration of each step. Type 5 is special: it can have **two** answers, which he knew.

````python
# Al-Khwarizmi's rules for "squares and roots equal numbers" and "squares and numbers equal roots"
def type4(b, c):
    '''x^2 + b*x = c   (positive solution)'''
    half = b / 2
    print(f"Equation: x² + {b:g}x = {c:g}")
    print(f"  1) Half of the roots: {b:g} / 2 = {half:g}")
    print(f"  2) Square it: {half:g} × {half:g} = {half * half:g}")
    print(f"  3) Add to the number: {c:g} + {half * half:g} = {c + half * half:g}")
    root = math.sqrt(c + half * half)
    print(f"  4) Square root: √{c + half * half:g} = {root:g}")
    x = root - half
    print(f"  5) Subtract the half: {root:g} − {half:g} = {x:g}   →  x = {x:g}")
    return [x]

def type5(b, c):
    '''x^2 + c = b*x   (can have two positive solutions)'''
    half = b / 2
    print(f"Equation: x² + {c:g} = {b:g}x")
    print(f"  1) Half of the roots: {b:g} / 2 = {half:g}")
    print(f"  2) Square it: {half:g} × {half:g} = {half * half:g}")
    print(f"  3) Subtract the number: {half * half:g} − {c:g} = {half * half - c:g}")
    root = math.sqrt(half * half - c)
    print(f"  4) Square root: √{half * half - c:g} = {root:g}")
    answers = [half - root, half + root]
    print(f"  5) Half minus it, and half plus it: {half:g} − {root:g} = {answers[0]:g}  and  {half:g} + {root:g} = {answers[1]:g}")
    return answers

for solve, b, c in [(type4, 10, 39), (type5, 10, 21)]:
    answers = solve(b, c)
    for x in answers:                                    # always verify!
        assert abs((x * x + b * x - c) if solve is type4 else (x * x + c - b * x)) < 1e-9
    print("  ✅ checked by substitution\n")
````

## ⚖️ Step 7: Al-jabr and al-muqabala, rebuilt
> 💡 The name of his book comes from two operations:
> - **Al-jabr** (*completion*): if a side has a **subtracted** amount, **add** that amount to both sides so only positive terms remain.
> - **Al-muqabala** (*balancing*): if the same kind of term appears on **both** sides, **cancel** the smaller one from both.
>
> Then he **reduces to one square** (divides so `x²` has coefficient 1) and recognises one of the six types. Our code does exactly this. We store each side as `{power of x: amount}`.

````python
# al-jabr, al-muqabala and the reduction, applied to any equation
NAMES = {2: "x²", 1: "x", 0: ""}

def fmt(side):
    parts = []
    for power in sorted(side, reverse=True):
        amount = side[power]
        if amount == 0:
            continue
        coef = "" if (amount == 1 and power > 0) else f"{amount:g}"
        parts.append(f"{coef}{NAMES[power]}")
    return " + ".join(parts) if parts else "0"

def show(left, right, note):
    print(f"  {note:<28} {fmt(left)}  =  {fmt(right)}")

def clean(side):
    return {p: a for p, a in side.items() if abs(a) > 1e-12}

def al_jabr(left, right):
    '''Completion: remove subtracted terms by adding them to both sides.'''
    left, right = dict(left), dict(right)
    for side in (left, right):
        for power, amount in list(side.items()):
            if amount < 0:
                for target in (left, right):
                    target[power] = target.get(power, 0) - amount
    return clean(left), clean(right)

def al_muqabala(left, right):
    '''Balancing: cancel equal kinds of terms from both sides.'''
    left, right = dict(left), dict(right)
    for power in set(left) & set(right):
        smaller = min(left[power], right[power])
        left[power] -= smaller; right[power] -= smaller
    return clean(left), clean(right)

def reduce_to_one_square(left, right):
    a = left.get(2, 0) - right.get(2, 0)
    if a in (0, 1):
        return left, right
    return {p: v / a for p, v in left.items()}, {p: v / a for p, v in right.items()}

def classify_and_solve(left, right):
    '''Recognise the type from the signs (everything moved to one side) and apply his rule.'''
    a = left.get(2, 0) - right.get(2, 0)
    b = left.get(1, 0) - right.get(1, 0)
    c = left.get(0, 0) - right.get(0, 0)
    if a < 0:
        a, b, c = -a, -b, -c
    if a == 1 and b > 0 and c < 0:   return "type 4: squares and roots equal numbers", type4(b, -c)
    if a == 1 and b < 0 and c > 0:   return "type 5: squares and numbers equal roots", type5(-b, c)
    if a == 1 and b < 0 and c == 0:  return "type 1: squares equal roots", [-b]
    if a == 1 and b == 0 and c < 0:  return "type 2: squares equal numbers", [math.sqrt(-c)]
    if a == 0 and b != 0:            return "type 3: roots equal numbers", [-c / b]
    if a == 1 and b < 0 and c < 0:   return "type 6: roots and numbers equal squares", [(-b) / 2 + math.sqrt(((-b) / 2) ** 2 - c)]
    raise ValueError("not one of the six types")

def solve_equation(left, right, title):
    print(f"\n=== {title} ===")
    show(left, right, "Given:")
    left, right = al_jabr(left, right);              show(left, right, "After al-jabr (completion):")
    left, right = al_muqabala(left, right);          show(left, right, "After al-muqabala (balancing):")
    left, right = reduce_to_one_square(left, right); show(left, right, "Reduced to one square:")
    kind, answers = classify_and_solve(left, right)
    print(f"  → recognised {kind}\n  → answer(s): {[round(x, 6) for x in answers]}")
    return answers

# 1) needs al-muqabala (the 29 cancels):  50 + x² = 29 + 10x
a1 = solve_equation({0: 50, 2: 1}, {0: 29, 1: 10}, "50 + x² = 29 + 10x")
assert sorted(a1) == [3.0, 7.0]
# 2) needs al-jabr and the reduction:  x² = 40x − 4x²
a2 = solve_equation({2: 1}, {1: 40, 2: -4}, "x² = 40x − 4x²")
assert a2 == [8.0]
````

## 🏎️ Step 8: Which algorithm is better? Count the steps
> 💡 The same job can be done by very different algorithms. We **count the steps** to compare them.
>
> **Job 1: greatest common divisor** (the biggest number that divides both).
> - *Naive:* try numbers from the smaller one downwards until one divides both.
> - *Euclid (about 300 BCE):* replace `(a, b)` by `(b, a mod b)` until `b` is 0.
>
> **Job 2: find a number in a sorted list.**
> - *Linear search:* check one by one.
> - *Binary search:* check the middle, then discard half of the list each time.

````python
# Count the steps of two algorithms for each job
def gcd_naive_steps(a, b):
    steps = 0
    for d in range(min(a, b), 0, -1):
        steps += 1
        if a % d == 0 and b % d == 0:
            return d, steps

def gcd_euclid_steps(a, b):
    steps = 0
    while b:
        a, b = b, a % b
        steps += 1
    return a, steps

def linear_search_steps(sorted_list, target):
    for steps, value in enumerate(sorted_list, start=1):
        if value == target:
            return steps

def binary_search_steps(sorted_list, target):
    low, high, steps = 0, len(sorted_list) - 1, 0
    while low <= high:
        steps += 1
        mid = (low + high) // 2
        if sorted_list[mid] == target:
            return steps
        low, high = (mid + 1, high) if sorted_list[mid] < target else (low, mid - 1)

rng = random.Random(5)
digits_range = [2, 3, 4, 5, 6]
gcd_rows = []
for k in digits_range:
    pairs = [(rng.randint(10 ** (k - 1), 10 ** k - 1), rng.randint(10 ** (k - 1), 10 ** k - 1)) for _ in range(10)]
    naive = np.mean([gcd_naive_steps(a, b)[1] for a, b in pairs])
    euclid = np.mean([gcd_euclid_steps(a, b)[1] for a, b in pairs])
    assert all(gcd_naive_steps(a, b)[0] == gcd_euclid_steps(a, b)[0] == math.gcd(a, b) for a, b in pairs)
    gcd_rows.append({"digits": k, "naive (avg steps)": round(naive), "Euclid (avg steps)": round(euclid, 1)})

sizes = [10, 100, 1_000, 10_000, 100_000]
search_rows = []
for n in sizes:
    data = list(range(n))
    targets = [rng.randrange(n) for _ in range(40)]
    search_rows.append({"list size": n,
                        "linear (avg steps)": round(np.mean([linear_search_steps(data, t) for t in targets])),
                        "binary (avg steps)": round(np.mean([binary_search_steps(data, t) for t in targets]), 1)})
gcd_table, search_table = pd.DataFrame(gcd_rows), pd.DataFrame(search_rows)
print("Greatest common divisor (numbers with k digits):\n", gcd_table.to_string(index=False))
print("\nSearching a sorted list:\n", search_table.to_string(index=False))

fig, axes = plt.subplots(1, 2, figsize=(12, 4.4))
axes[0].plot(gcd_table["digits"], gcd_table["naive (avg steps)"], marker="o", lw=3, color=RED, label="naive")
axes[0].plot(gcd_table["digits"], gcd_table["Euclid (avg steps)"], marker="o", lw=3, color=GREEN, label="Euclid")
axes[0].set_yscale("log"); axes[0].set_xlabel("digits in each number"); axes[0].set_ylabel("steps (log scale)")
axes[0].set_title("Greatest common divisor", color=DARK, fontweight="bold"); axes[0].legend(); axes[0].grid(alpha=0.3)
axes[1].plot(search_table["list size"], search_table["linear (avg steps)"], marker="o", lw=3, color=RED, label="linear")
axes[1].plot(search_table["list size"], search_table["binary (avg steps)"], marker="o", lw=3, color=GREEN, label="binary")
axes[1].set_xscale("log"); axes[1].set_yscale("log"); axes[1].set_xlabel("list size"); axes[1].set_ylabel("steps (log scale)")
axes[1].set_title("Searching a sorted list", color=DARK, fontweight="bold"); axes[1].legend(); axes[1].grid(alpha=0.3)
plt.tight_layout()
plt.show()
````

## 🌀 Step 9: Does it always stop? The Collatz puzzle
> 💡 One property of an algorithm is **finiteness**: it must stop. Try this tiny procedure:
> *start with any positive number. If it is even, halve it. If it is odd, triple it and add 1. Repeat until you reach 1.*
>
> It stops for **every number anyone has ever tried**. Yet **nobody has proved** it always stops. This is the **Collatz conjecture** (1937), and it is still open. It shows that even a simple recipe can hide deep questions.

````python
# The Collatz procedure: count the steps to reach 1
def collatz_steps(n):
    steps = 0
    while n != 1:
        n = n // 2 if n % 2 == 0 else 3 * n + 1
        steps += 1
    return steps

print("Path of 27:", end=" ")
n, path = 27, [27]
while n != 1:
    n = n // 2 if n % 2 == 0 else 3 * n + 1
    path.append(n)
print(f"{len(path) - 1} steps, highest value reached = {max(path):,}")

numbers = list(range(1, 101))
steps_list = [collatz_steps(k) for k in numbers]
plt.figure(figsize=(9, 4.4))
plt.bar(numbers, steps_list, color=TEAL, edgecolor="white")
plt.xlabel("starting number"); plt.ylabel("steps to reach 1")
plt.title("Collatz: the number of steps jumps around unpredictably", color=DARK, fontweight="bold")
plt.grid(axis="y", alpha=0.3)
plt.show()
````

## ⚠️ Limitations of algorithms

| Limitation | What it means | Who looked into it |
|---|---|---|
| ❌ **Needs exact steps** | If a task cannot be described precisely, there is no algorithm for it | Later: machine learning learns the steps from examples |
| ❌ **Speed can explode** | A correct algorithm may take longer than the age of the universe | Complexity theory (1960s and 1970s) |
| ❌ **Some problems have no algorithm at all** | Turing proved (1936) that no algorithm can decide whether *every* program stops | Alan Turing |
| ❌ **Correctness must be proven** | A recipe that "looks right" can still fail on rare inputs | Program verification |

### 🧬 What it gave us
- The very idea of a **general procedure** that anyone (or any machine) can follow
- **Positional numbers and zero**, the base of all modern calculation
- **Algebra**, the language of every formula in AI
- The word we use every day: **algorithm**
